```python
df["Cuisines"].isnull().sum() / len(df) * 100
```

```python
# 7. Remove missing cuisine values
df_cuisines = df.dropna(subset=["Cuisines"])
df_cuisines.shape
```

```python
# 8. Split the cuisines
cuisine_series = df_cuisines["Cuisines"].str.split(", ")
cuisine_series.head()
```

```python
# 9. Flatten all cuisine lists and count them
all_cuisines = cuisine_series.explode()
cuisine_counts = all_cuisines.value_counts()
cuisine_counts.head(10)
```

```python
# percentage
top_3_cuisines = cuisine_counts.head(3)
percentage = (top_3_cuisines / len(df_cuisines)) * 100
percentage
```

```python
# 10. Visualization
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(top_3_cuisines.index, top_3_cuisines.values)

plt.title("Top 3 Most Common Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")

plt.show()
```
