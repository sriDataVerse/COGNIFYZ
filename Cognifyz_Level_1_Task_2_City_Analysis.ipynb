```python
#1. Import libraries
import pandas as pd
```

```python
# 2. Load the dataset
df = pd.read_csv("C:\\Users\\sksri\\OneDrive\\Documents\\Cognifyz\\Dataset .csv")
df.head()
```

```python
# 3. Check the City column
df["City"].head()
```

```python
# 4. Check missing values in the Cuisines column
df["City"].isnull().sum()
```

```python
# 5. Count restaurants in each city
city_counts = df["City"].value_counts()
city_counts.head(10)
```

```python
# 6. Calculating the average rating for restaurants in each city
city_avg_rating = df.groupby("City")["Aggregate rating"].mean()
city_avg_rating.sort_values(ascending=False).head(10)
```

```python
# 7. Visualization
import matplotlib.pyplot as plt

top_10_cities = city_avg_rating.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.bar(top_10_cities.index, top_10_cities.values)

plt.title("Top 10 Cities by Average Restaurant Rating")
plt.xlabel("City")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)

plt.show()
```

```python
 ## Conclusion

New Delhi has the highest number of restaurants in the dataset, with 5,473 restaurants.
Based on the average restaurant ratings, Inner City has the highest average rating of 4.90, followed by Quezon City with an average rating of 4.80.
Therefore, New Delhi is the city with the highest restaurant presence, while Inner City has the highest average restaurant rating in the dataset.
```
