```python
# 1. Import libraries
import pandas as pd
```

```python
# 2. Load the dataset
df = pd.read_csv("C:\\Users\\sksri\\OneDrive\\Documents\\Cognifyz\\Dataset .csv")
df.head()
```

```python
df["Has Online delivery"].value_counts()
```

```python
# 3. Calculate percentage
online_delivery_percentage = (df["Has Online delivery"].value_counts() / len(df)) * 100
online_delivery_percentage
```

```python
# 4. Compare average ratings
rating_comparison = df.groupby("Has Online delivery")["Aggregate rating"].mean()
rating_comparison
```

```python
# 5. Visualization
import matplotlib.pyplot as plt

rating_comparison.plot(kind="bar", figsize=(6,5))

plt.title("Average Rating by Online Delivery Availability")
plt.xlabel("Has Online Delivery")
plt.ylabel("Average Rating")

plt.show()
```

```python
## Conclusion

#The analysis shows that 25.66% of restaurants offer online delivery, while 74.34% do not offer online delivery.
#Restaurants with online delivery have an average rating of 3.25, whereas restaurants without online delivery have an average rating of 2.47.
#This indicates that restaurants providing online delivery tend to receive higher customer ratings compared to those that do not offer online delivery.
```
