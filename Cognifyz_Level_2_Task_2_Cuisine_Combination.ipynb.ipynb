```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 2. Load the dataset
df = pd.read_csv("C:\\Users\\sksri\\OneDrive\\Documents\\Cognifyz\\Dataset .csv")
df.head()
```

```python
# 3. Check for missing values 
df["Cuisines"].isnull().sum()
```

```python
# 4. Remove Missing Cuisine Values
df_cuisines = df.dropna(subset=["Cuisines"])
df_cuisines.shape
```

```python
# 5. Finding the Most Common Cuisine Combinations
cuisine_combinations = df_cuisines["Cuisines"].value_counts()
cuisine_combinations.head(10)
```

```python
# 6. Find Which Cuisine Combinations Have Higher Ratings
top_combinations = cuisine_combinations.head(10).index

rating_by_cuisine = (
    df_cuisines[df_cuisines["Cuisines"].isin(top_combinations)]
    .groupby("Cuisines")["Aggregate rating"]
    .mean()
    .sort_values(ascending=False)
)

rating_by_cuisine
```

```python
# 7. Visualization
plt.figure(figsize=(10,5))

rating_by_cuisine.plot(kind="bar")

plt.title("Average Rating by Cuisine Combination")
plt.xlabel("Cuisine Combination")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)
plt.show()
```

```python
## Conclusion

#The analysis of cuisine combinations shows that North Indian is the most common cuisine category, followed by North Indian, Chinese and Chinese cuisines.
#Among the top cuisine combinations, Cafe and North Indian, Mughlai received the highest average ratings, indicating better customer satisfaction compared to other cuisine categories.
#North Indian cuisine alone had the lowest average rating among the top cuisine categories.
#Overall, the results suggest that certain cuisine combinations tend to achieve higher customer ratings than single-cuisine categories.
```
