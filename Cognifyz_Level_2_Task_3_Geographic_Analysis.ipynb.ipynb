```python
# 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 2. Load the Dataset
df = pd.read_csv("C:\\Users\\sksri\\OneDrive\\Documents\\Cognifyz\\Dataset .csv")
df.head()
```

```python
# 3. Check the Geographic Columns
df[["Latitude", "Longitude"]].head()
```

```python
# 4. Check for missing value
df[["Latitude", "Longitude"]].isnull().sum()
```

```python
# 5. Visualization

#Create a Geographic Scatter Plot

#Latitude → Y-axis
#Longitude → X-axis

plt.figure(figsize=(10,6))

plt.scatter(df["Longitude"], df["Latitude"], alpha=0.5)

plt.title("Restaurant Locations")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.show()
```

```python
# 6. Find the Geographic Center
avg_latitude = df["Latitude"].mean()
avg_longitude = df["Longitude"].mean()

print("Average Latitude:", avg_latitude)
print("Average Longitude:", avg_longitude)
```

```python
## Conclusion

#The geographic analysis was performed using the latitude and longitude coordinates of restaurants in the dataset.
#No missing values were found in the geographic coordinates, allowing complete location analysis.
#The scatter plot revealed that restaurants are concentrated in specific geographic regions rather than being uniformly distributed.
#The average latitude and longitude of all restaurants were 25.85 and 64.13 respectively, representing the approximate geographic center of the dataset.
#Overall, the analysis indicates that restaurant locations are clustered in certain areas, suggesting higher restaurant density in specific regions.
```
