```python
# 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 2. Load the Dataset
df = pd.read_csv("C:\\Users\\sksri\\OneDrive\\Documents\\Cognifyz\\Dataset .csv")
df.head()
```

```python
# 3. Finding the Restaurant Chains
restaurant_counts = df["Restaurant Name"].value_counts()
restaurant_counts.head(10)
```

```python
# 4. Analyzing the Ratings of Top Chains
top_chains = restaurant_counts.head(10).index

chain_ratings = (
    df[df["Restaurant Name"].isin(top_chains)]
    .groupby("Restaurant Name")["Aggregate rating"]
    .mean()
    .sort_values(ascending=False)
)

chain_ratings
```

```python
# 5. Analyzing the Popularity Using Votes
chain_votes = (
    df[df["Restaurant Name"].isin(top_chains)]
    .groupby("Restaurant Name")["Votes"]
    .mean()
    .sort_values(ascending=False)
)

chain_votes
```

```python
# 6. Visualization
plt.figure(figsize=(10,5))

chain_ratings.plot(kind="bar")

plt.title("Average Ratings of Top Restaurant Chains")
plt.xlabel("Restaurant Chain")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)
plt.show()
```

```python
## Conclusion

#The analysis identified several restaurant chains operating across multiple locations in the dataset.
#Cafe Coffee Day, Domino's Pizza, and Subway were the most widespread chains based on the number of outlets.
#Among the top restaurant chains, Barbeque Nation achieved the highest average rating of 4.35 and also received the highest average number of votes, indicating strong customer satisfaction and popularity.
#The results suggest that highly rated restaurant chains tend to attract greater customer engagement and positive feedback.
#Overall, Barbeque Nation emerged as the strongest-performing restaurant chain in terms of both ratings and popularity.
```
